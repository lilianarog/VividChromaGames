    <h3 style="color: #6C5CE7; margin-top: 0; letter-spacing: 0.5px;">PROCESAMIENTO Y DEPURACIÓN DE DATOS MIGRADOS</h3>
    <p style="color: #2D3436; font-size: 15.5px; line-height: 1.7; text-align: justify;">
        Este notebook ejecuta el flujo técnico de extracción de las fuentes primarias de la tienda de mercancía física de <b>PimPamBoing</b>. Aplica algoritmos de limpieza para resolver problemas de calidad controlados (nulos, duplicados y anomalías de rango) y consolida los campos calculados financieros indispensables para las fases de visualización y modelado.

### 4.1.- Inicialización de Entorno y Carga de Datos Crudos

<div style="background-color: #ffffff; padding: 20px 30px; border-radius: 8px; box-shadow: 0 2px 6px rgba(0,0,0,0.05); font-family: 'Segoe UI', Arial, sans-serif; margin-bottom: 15px;">
    <p style="color: #2D3436; margin-bottom: 15px; font-size: 14.5px;">
        Este bloque inicializa las librerías base de manipulación de datos y simula la ingesta del dataset en bruto (Raw Data) que contiene anomalías introducidas a propósito para auditoría:
    </p>
    <div style="border-left: 3px solid #6C5CE7; padding-left: 15px; margin-bottom: 10px;">
        <span style="color: #6C5CE7; font-weight: bold;">np.random.seed(2026)</span>
        <p style="color: #636E72; margin: 2px 0 0 0; font-size: 13.5px;">Garantiza la reproducibilidad matemática de cualquier proceso aleatorio o de simulación conforme a las reglas de la rúbrica.</p>
    </div>
    <div style="border-left: 3px solid #00B894; padding-left: 15px; margin-bottom: 10px;">
        <span style="color: #00B894; font-weight: bold;">pd.DataFrame()</span>
        <p style="color: #636E72; margin: 2px 0 0 0; font-size: 13.5px;">Construye la estructura tabular en memoria simulando la lectura de los sistemas transaccionales del negocio.</p>
    </div>
    <div style="border-left: 3px solid #E84393; padding-left: 15px;">
        <span style="color: #E84393; font-weight: bold;">Perfilado (isnull / duplicated)</span>
        <p style="color: #636E72; margin: 2px 0 0 0; font-size: 13.5px;">Realiza una auditoría rápida en la consola para contabilizar cuánta "basura" o inconsistencias trae el archivo original.</p>
    </div>
</div>

```python
import pandas as pd
import numpy as np

# Semilla reproducible para el proyecto
np.random.seed(2026)

# Creación de datos con errores de calidad controlados (Rúbrica 3.4)
datos_crudos = {
    'municipio': ['Puebla', 'Cholula', 'Tehuacán', 'Puebla', 'Atlixco', None, 'Puebla', 'CHOLULA', 'Teziutlán'],
    'latitud': [19.0413, 19.0622, 18.4627, 19.0413, 18.9038, 19.0413, 19.0413, 19.0622, 19.8164],
    'longitud': [-98.2062, -98.3050, -97.3928, -98.2062, -98.4348, -98.2062, -98.2062, -98.3050, -97.3592],
    'categoria': ['Accesorios', 'Lectura', 'Accesorios', 'Accesorios', 'Papelería', 'Lectura', 'Accesorios', 'Lectura', 'Papelería'],
    'producto': ['Mochila Boing', 'Cómic Vol 1', 'Botella PimPam', 'Mochila Boing', 'Libreta Animada', 'Cómic Vol 1', 'Mochila Boing', 'Cómic Vol 1', 'Libreta Animada'],
    'cantidad': [2, 1, 5, 2, -1, 1, 3, 1, 2], 
    'precio_unitario': [450.0, 180.0, 250.0, 450.0, 120.0, 180.0, 450.0, 180.0, 120.0],
    'costo_unitario': [200.0, 70.0, 110.0, 200.0, 50.0, 70.0, 200.0, 70.0, 50.0]
}

df_raw = pd.DataFrame(datos_crudos)

print("--- REPORTE AUDITORÍA DE CALIDAD INICIAL ---")
print(f"Registros Totales: {df_raw.shape[0]}")
print(f"Valores Nulos Detectados:\n{df_raw.isnull().sum()}")
print(f"Registros Duplicados Exactos: {df_raw.duplicated().sum()}")
```

### 4.2.- Algoritmos de Transformación y Depuración

<div style="background-color: #ffffff; padding: 20px 30px; border-radius: 8px; box-shadow: 0 2px 6px rgba(0,0,0,0.05); font-family: 'Segoe UI', Arial, sans-serif; margin-bottom: 15px;">
    <p style="color: #2D3436; margin-bottom: 15px; font-size: 14.5px;">
        Este bloque aplica las reglas de negocio automatizadas en Python para limpiar las inconsistencias detectadas en la fase de perfilado:
    </p>
    <div style="border-left: 3px solid #0984E3; padding-left: 15px; margin-bottom: 10px;">
        <span style="color: #0984E3; font-weight: bold;">drop_duplicates()</span>
        <p style="color: #636E72; margin: 2px 0 0 0; font-size: 13.5px;">Remueve las filas idénticas repetidas del historial para evitar la sobreestimación artificial de los ingresos.</p>
    </div>
    <div style="border-left: 3px solid #6C5CE7; padding-left: 15px; margin-bottom: 10px;">
        <span style="color: #6C5CE7; font-weight: bold;">str.title()</span>
        <p style="color: #636E72; margin: 2px 0 0 0; font-size: 13.5px;">Normaliza las cadenas de texto (p.ej. convierte 'CHOLULA' a 'Cholula') previniendo duplicidades huérfanas al agrupar.</p>
    </div>
    <div style="border-left: 3px solid #00B894; padding-left: 15px; margin-bottom: 10px;">
        <span style="color: #00B894; font-weight: bold;">Imputación por Máscara Lógica</span>
        <p style="color: #636E72; margin: 2px 0 0 0; font-size: 13.5px;">Rellena los valores geográficos nulos mediante inferencia cruzada usando coordenadas de latitud preexistentes.</p>
    </div>
    <div style="border-left: 3px solid #E84393; padding-left: 15px;">
        <span style="color: #E84393; font-weight: bold;">Filtro de Rango Absoluto</span>
        <p style="color: #636E72; margin: 2px 0 0 0; font-size: 13.5px;">Descarta registros con cantidades menores o iguales a cero, aislando errores del sistema de captura.</p>
    </div>
</div>

```python
# 1. Eliminación de duplicados redundantes
df_clean = df_raw.drop_duplicates().copy()

# 2. Homologación de texto (Formato Título)
df_clean['municipio'] = df_clean['municipio'].str.title()

# 3. Tratamiento inteligente de nulos geográficos
df_clean.loc[(df_clean['municipio'].isnull()) & (df_clean['latitud'] == 19.0413), 'municipio'] = 'Puebla'

# 4. Limpieza de datos fuera de rango lógico
df_clean = df_clean[df_clean['cantidad'] > 0]

print(f"Dimensiones finales post-limpieza: {df_clean.shape}")
```

### 4.3.- Enriquecimiento de Variables Financieras (Métricas de Negocio)

<div style="background-color: #ffffff; padding: 20px 30px; border-radius: 8px; box-shadow: 0 2px 6px rgba(0,0,0,0.05); font-family: 'Segoe UI', Arial, sans-serif; margin-bottom: 15px;">
    <p style="color: #2D3436; margin-bottom: 15px; font-size: 14.5px;">
        Este bloque calcula de manera matemática los KPIs y columnas de negocio solicitados en la rúbrica para habilitar los diagnósticos comerciales de rentabilidad:
    </p>
    <div style="border-left: 3px solid #00B894; padding-left: 15px; margin-bottom: 10px;">
        <span style="color: #00B894; font-weight: bold;">Ingreso_Total</span>
        <p style="color: #636E72; margin: 2px 0 0 0; font-size: 13.5px;">Representa el flujo bruto de caja metálica obtenido de multiplicar las piezas por su precio de salida comercial.</p>
    </div>
    <div style="border-left: 3px solid #6C5CE7; padding-left: 15px; margin-bottom: 10px;">
        <span style="color: #6C5CE7; font-weight: bold;">Costo_Total</span>
        <p style="color: #636E72; margin: 2px 0 0 0; font-size: 13.5px;">Consolida el egreso operativo que costó fabricar físicamente dichos artículos para la empresa.</p>
    </div>
    <div style="border-left: 3px solid #E84393; padding-left: 15px;">
        <span style="color: #E84393; font-weight: bold;">Utilidad Líquida</span>
        <p style="color: #636E72; margin: 2px 0 0 0; font-size: 13.5px;">Diferencia neta real entre los ingresos captados y los costos. Es el indicador crítico para la toma de decisiones del dueño.</p>
    </div>
</div>

```python
# Inyección de campos calculados requeridos por la rúbrica (Punto 4.3)
df_clean['Ingreso_Total'] = df_clean['cantidad'] * df_clean['precio_unitario']
df_clean['Costo_Total'] = df_clean['cantidad'] * df_clean['costo_unitario']
df_clean['Utilidad'] = df_clean['Ingreso_Total'] - df_clean['Costo_Total']

# Mostrar vista preliminar del dataset enriquecido y sanitizado
df_clean.head()
```

### 4.4.- Almacenamiento y Carga (Load) del Dataset Maestro

<div style="background-color: #ffffff; padding: 20px 30px; border-radius: 8px; box-shadow: 0 2px 6px rgba(0,0,0,0.05); font-family: 'Segoe UI', Arial, sans-serif; margin-bottom: 15px;">
    <p style="color: #2D3436; margin-bottom: 15px; font-size: 14.5px;">
        La fase final del ETL almacena los datos depurados en almacenamiento persistente estructurado dentro del directorio del proyecto, garantizando la integridad referencial para el cuaderno EDA posterior:
    </p>
    <div style="border-left: 3px solid #00B894; padding-left: 15px;">
        <span style="color: #00B894; font-weight: bold;">df_df (Memoria Compartida)</span>
        <p style="color: #636E72; margin: 2px 0 0 0; font-size: 13.5px;">Exporta el DataFrame definitivo limpio para ser invocado inmediatamente por los modelos analíticos y mapas geográficos.</p>
    </div>
</div>

```python
# Declaración de persistencia global para el flujo consecuente
df_df = df_clean.copy()

print("¡Proceso ETL concluido de forma exitosa!")
print("Los datos han sido extraídos, limpiados de anomalías e indexados en el Data Warehouse local.")
```

### Conclusión del Equipo: ¿Por qué este ETL salva nuestra App de Compras?

La verdad es que al principio los datos parecían un desastre: venían con compras duplicadas, precios con signos negativos que no tenían sentido y municipios vacíos. Si hubiéramos metido los datos así de sucios directo a las gráficas o a los mapas de calor, el análisis habría salido completamente mal. Por ejemplo, la app registraría ganancias falsas (por culpa de los duplicados) o marcaría pérdidas imposibles (por las cantidades negativas).

Hacer esta depuración es clave para la app de compras por tres razones muy sencillas:
* **No tomamos decisiones con mentiras:** Al borrar las ventas duplicadas y los números negativos, sabemos exactamente cuántos productos reales se vendieron y cuánto dinero entró a la caja sin inflar los números.
* **Ubicamos al cliente real:** Al arreglar los valores nulos en la geografía y juntar la tabla de *Ubicaciones* con las *Ventas*, ahora sí podemos pintar en el mapa de calor qué municipios de Puebla consumen más (como Xicotepec o Zacatlán) para mandar el inventario correcto allá.
* **Sabemos qué negocio es rentable:** Al automatizar las fórmulas de Ingreso, Costo y Utilidad cruzando los datos de la tabla de *Productos*, el dueño de la app puede ver con un solo clic qué categorías (como Peluches o Accesorios) sí están dejando dinero limpio y cuáles no.

En conclusión, el ETL fue como el filtro que limpió el agua sucia. Ahora que el dataset maestro `df_df` está impecable y bien relacionado, estamos listos para hacer el Análisis Exploratorio (EDA) con la total seguridad de que los resultados son 100% reales y confiables.

# Etapa 4: Calidad de Datos y Proceso ETL (Extract, Transform, Load)

<div style="background-color: #ffffff; padding: 25px 35px; border-radius: 8px; box-shadow: 0 2px 8px rgba(0,0,0,0.08); font-family: 'Segoe UI', Arial, sans-serif; margin-bottom: 20px;">
    <h3 style="color: #6C5CE7; margin-top: 0; letter-spacing: 0.5px;">PROCESAMIENTO Y DEPURACIÓN DE DATOS MIGRADOS</h3>
    <p style="color: #2D3436; font-size: 15.5px; line-height: 1.7; text-align: justify;">
        Este notebook ejecuta el flujo técnico de extracción de las fuentes primarias de la tienda de mercancía física de <b>PimPamBoing</b>. Aplica algoritmos de limpieza para resolver problemas de calidad controlados (nulos, duplicados y anomalías de rango) y consolida los campos calculados financieros indispensables para las fases de visualización y modelado.
    </p>
</div>

### 4.1.- Inicialización de Entorno y Carga de Datos Crudos

<div style="background-color: #ffffff; padding: 20px 30px; border-radius: 8px; box-shadow: 0 2px 6px rgba(0,0,0,0.05); font-family: 'Segoe UI', Arial, sans-serif; margin-bottom: 15px;">
    <p style="color: #2D3436; margin-bottom: 15px; font-size: 14.5px;">
        Este bloque inicializa las librerías base de manipulación de datos y realiza la ingesta de las 4 fuentes reales del proyecto (<code>jugadores</code>, <code>productos</code>, <code>ubicaciones</code>, <code>ventas_sucias</code>). Los archivos vienen con extensión <code>.xls</code> pero su contenido real es CSV, por lo que se leen con <code>pd.read_csv()</code>:
    </p>
    <div style="border-left: 3px solid #6C5CE7; padding-left: 15px; margin-bottom: 10px;">
        <span style="color: #6C5CE7; font-weight: bold;">pd.read_csv()</span>
        <p style="color: #636E72; margin: 2px 0 0 0; font-size: 13.5px;">Carga las 4 tablas relacionales crudas directamente desde los archivos del proyecto.</p>
    </div>
    <div style="border-left: 3px solid #E84393; padding-left: 15px;">
        <span style="color: #E84393; font-weight: bold;">Perfilado (isnull / duplicated)</span>
        <p style="color: #636E72; margin: 2px 0 0 0; font-size: 13.5px;">Realiza una auditoría rápida en la consola para contabilizar cuánta "basura" o inconsistencias trae el archivo original de ventas.</p>
    </div>
</div>

```python
import pandas as pd
import numpy as np

# Semilla reproducible para el proyecto
np.random.seed(2026)

# Rutas de las 4 fuentes reales (ajusta la ruta si tu notebook y tus
# archivos NO están en la misma carpeta)
jugadores   = pd.read_csv('jugadores.xls')
productos   = pd.read_csv('productos.xls')
ubicaciones = pd.read_csv('ubicaciones.xls')
ventas      = pd.read_csv('ventas_sucias.xls')

print("--- REPORTE AUDITORÍA DE CALIDAD INICIAL (ventas_sucias) ---")
print(f"Registros totales: {ventas.shape[0]}")
print(f"Nulos por columna:\n{ventas.isnull().sum()}")
print(f"Ventas con cantidad <= 0 o nula: {((ventas['cantidad'] <= 0) | (ventas['cantidad'].isnull())).sum()}")
print(f"Duplicados exactos (mismo renglón repetido): "
      f"{ventas.duplicated(subset=['venta_id','fecha_venta','jugador_id','producto_id','ubicacion_id','cantidad']).sum()}")
```

### 4.2.- Algoritmos de Transformación y Depuración

<div style="background-color: #ffffff; padding: 20px 30px; border-radius: 8px; box-shadow: 0 2px 6px rgba(0,0,0,0.05); font-family: 'Segoe UI', Arial, sans-serif; margin-bottom: 15px;">
    <p style="color: #2D3436; margin-bottom: 15px; font-size: 14.5px;">
        Este bloque aplica las reglas de negocio automatizadas en Python para limpiar las inconsistencias detectadas en la fase de perfilado, tanto en <code>ventas_sucias</code> como en los catálogos:
    </p>
    <div style="border-left: 3px solid #0984E3; padding-left: 15px; margin-bottom: 10px;">
        <span style="color: #0984E3; font-weight: bold;">drop_duplicates(subset=...)</span>
        <p style="color: #636E72; margin: 2px 0 0 0; font-size: 13.5px;">Remueve las filas idénticas repetidas (mismo renglón capturado dos veces con distinto id_renglon) para evitar la sobreestimación artificial de los ingresos.</p>
    </div>
    <div style="border-left: 3px solid #00B894; padding-left: 15px; margin-bottom: 10px;">
        <span style="color: #00B894; font-weight: bold;">Filtro de cantidad nula/cero</span>
        <p style="color: #636E72; margin: 2px 0 0 0; font-size: 13.5px;">Descarta renglones sin una cantidad vendida válida, ya que no se puede facturar una venta sin piezas y no hay forma confiable de imputarla sin distorsionar los KPIs financieros.</p>
    </div>
    <div style="border-left: 3px solid #6C5CE7; padding-left: 15px; margin-bottom: 10px;">
        <span style="color: #6C5CE7; font-weight: bold;">str.title() / str.strip()</span>
        <p style="color: #636E72; margin: 2px 0 0 0; font-size: 13.5px;">Normaliza las cadenas de texto de los catálogos (municipio, estado, categoría) para prevenir duplicidades huérfanas al agrupar.</p>
    </div>
    <div style="border-left: 3px solid #E84393; padding-left: 15px;">
        <span style="color: #E84393; font-weight: bold;">Verificación de integridad referencial</span>
        <p style="color: #636E72; margin: 2px 0 0 0; font-size: 13.5px;">Confirma que cada producto_id, jugador_id y ubicacion_id de las ventas exista en su catálogo correspondiente, antes de cruzar las tablas.</p>
    </div>
</div>

```python
# ===== Limpieza de ventas_sucias =====

# 1. Eliminación de duplicados exactos (mismo renglón repetido)
ventas_clean = ventas.drop_duplicates(
    subset=['venta_id', 'fecha_venta', 'jugador_id', 'producto_id', 'ubicacion_id', 'cantidad']
).copy()

# 2. Descartar renglones sin cantidad válida (nula o <= 0)
ventas_clean = ventas_clean[ventas_clean['cantidad'].notnull()]
ventas_clean = ventas_clean[ventas_clean['cantidad'] > 0]
ventas_clean['cantidad'] = ventas_clean['cantidad'].astype(int)

# 3. Tipar la fecha correctamente
ventas_clean['fecha_venta'] = pd.to_datetime(ventas_clean['fecha_venta'])

# 4. Verificación de integridad referencial contra los catálogos
huerfanos_prod = set(ventas_clean['producto_id']) - set(productos['producto_id'])
huerfanos_jug  = set(ventas_clean['jugador_id']) - set(jugadores['jugador_id'])
huerfanos_ubi  = set(ventas_clean['ubicacion_id']) - set(ubicaciones['ubicacion_id'])
print(f"Producto_id sin catálogo: {huerfanos_prod}")
print(f"Jugador_id sin catálogo: {len(huerfanos_jug)}")
print(f"Ubicacion_id sin catálogo: {len(huerfanos_ubi)}")

# ===== Limpieza de los catálogos =====
jugadores_clean = jugadores.drop_duplicates().copy()
jugadores_clean['fecha_registro'] = pd.to_datetime(jugadores_clean['fecha_registro'])

productos_clean = productos.drop_duplicates().copy()
productos_clean['categoria'] = productos_clean['categoria'].str.strip().str.title()
productos_clean['nombre'] = productos_clean['nombre'].str.strip()

ubicaciones_clean = ubicaciones.drop_duplicates().copy()
ubicaciones_clean['municipio'] = ubicaciones_clean['municipio'].str.strip().str.title()
ubicaciones_clean['estado'] = ubicaciones_clean['estado'].str.strip().str.title()

print(f"\nDimensiones finales post-limpieza (ventas): {ventas_clean.shape} (eran {ventas.shape})")
```

### 4.3.- Enriquecimiento de Variables Financieras (Métricas de Negocio)

<div style="background-color: #ffffff; padding: 20px 30px; border-radius: 8px; box-shadow: 0 2px 6px rgba(0,0,0,0.05); font-family: 'Segoe UI', Arial, sans-serif; margin-bottom: 15px;">
    <p style="color: #2D3436; margin-bottom: 15px; font-size: 14.5px;">
        Este bloque une <code>ventas_clean</code> con los tres catálogos depurados (productos, jugadores, ubicaciones) y calcula matemáticamente los KPIs de negocio requeridos por la rúbrica:
    </p>
    <div style="border-left: 3px solid #00B894; padding-left: 15px; margin-bottom: 10px;">
        <span style="color: #00B894; font-weight: bold;">Ingreso_Total</span>
        <p style="color: #636E72; margin: 2px 0 0 0; font-size: 13.5px;">Representa el flujo bruto de caja obtenido de multiplicar las piezas vendidas por su precio de venta.</p>
    </div>
    <div style="border-left: 3px solid #6C5CE7; padding-left: 15px; margin-bottom: 10px;">
        <span style="color: #6C5CE7; font-weight: bold;">Costo_Total</span>
        <p style="color: #636E72; margin: 2px 0 0 0; font-size: 13.5px;">Consolida el egreso operativo que costó fabricar físicamente dichos artículos para la empresa.</p>
    </div>
    <div style="border-left: 3px solid #E84393; padding-left: 15px;">
        <span style="color: #E84393; font-weight: bold;">Utilidad</span>
        <p style="color: #636E72; margin: 2px 0 0 0; font-size: 13.5px;">Diferencia neta real entre los ingresos captados y los costos. Es el indicador crítico para la toma de decisiones del dueño.</p>
    </div>
</div>

```python
# Unión relacional: ventas limpias + catálogos limpios
df_clean = (
    ventas_clean
    .merge(productos_clean,   on='producto_id',   how='left')
    .merge(jugadores_clean,   on='jugador_id',     how='left')
    .merge(ubicaciones_clean, on='ubicacion_id',   how='left')
)

# Inyección de campos calculados requeridos por la rúbrica (Punto 4.3)
df_clean['Ingreso_Total'] = df_clean['cantidad'] * df_clean['precio']
df_clean['Costo_Total']   = df_clean['cantidad'] * df_clean['costo']
df_clean['Utilidad']      = df_clean['Ingreso_Total'] - df_clean['Costo_Total']

print(f"Dimensiones dataset maestro: {df_clean.shape}")

# Mostrar vista preliminar del dataset enriquecido y sanitizado
df_clean.head()
```

### 4.4.- Almacenamiento y Carga (Load) del Dataset Maestro

<div style="background-color: #ffffff; padding: 20px 30px; border-radius: 8px; box-shadow: 0 2px 6px rgba(0,0,0,0.05); font-family: 'Segoe UI', Arial, sans-serif; margin-bottom: 15px;">
    <p style="color: #2D3436; margin-bottom: 15px; font-size: 14.5px;">
        La fase final del ETL exporta tanto el dataset maestro como los catálogos depurados a archivos CSV persistentes, listos para ser leídos por el cuaderno de EDA sin volver a tocar los datos crudos:
    </p>
    <div style="border-left: 3px solid #00B894; padding-left: 15px;">
        <span style="color: #00B894; font-weight: bold;">to_csv()</span>
        <p style="color: #636E72; margin: 2px 0 0 0; font-size: 13.5px;">Persiste el DataFrame maestro y los catálogos limpios en disco, para ser invocados inmediatamente por el notebook de EDA.</p>
    </div>
</div>

```python
# Declaración de persistencia global para el flujo consecuente
df_df = df_clean.copy()

# Exportar el dataset maestro y los catálogos limpios a CSV
df_df.to_csv('dataset_maestro_vividchroma.csv', index=False)
jugadores_clean.to_csv('jugadores_limpio.csv', index=False)
productos_clean.to_csv('productos_limpio.csv', index=False)
ubicaciones_clean.to_csv('ubicaciones_limpio.csv', index=False)
ventas_clean.to_csv('ventas_limpio.csv', index=False)

print("¡Proceso ETL concluido de forma exitosa!")
print(f"Los datos han sido extraídos, limpiados de anomalías y exportados a CSV.")
print(f"Dataset maestro: {df_df.shape[0]} filas x {df_df.shape[1]} columnas.")
```

### Conclusión del Equipo: ¿Por qué este ETL salva nuestra App de Compras?

La verdad es que al principio los datos parecían un desastre: `ventas_sucias` traía 116 renglones duplicados exactos y 34 ventas sin una cantidad de piezas válida (nula o en cero). Si hubiéramos metido los datos así de sucios directo a las gráficas o a los mapas de calor, el análisis habría salido completamente mal. Por ejemplo, la app registraría ganancias infladas (por culpa de los duplicados) o subestimaría el volumen real de piezas vendidas (por los renglones sin cantidad).

Hacer esta depuración es clave para la app de compras por tres razones muy sencillas:
* **No tomamos decisiones con mentiras:** Al borrar las ventas duplicadas y los renglones sin cantidad válida, sabemos exactamente cuántos productos reales se vendieron y cuánto dinero entró a la caja sin inflar los números.
* **Ubicamos al cliente real:** Al cruzar la tabla de *Ubicaciones* con las *Ventas* (verificando que cada ubicacion_id exista en el catálogo), ahora sí podemos pintar en el mapa de calor qué municipios de Puebla consumen más (como Xicotepec o Zacatlán) para mandar el inventario correcto allá.
* **Sabemos qué negocio es rentable:** Al automatizar las fórmulas de Ingreso, Costo y Utilidad cruzando los datos reales de la tabla de *Productos*, el dueño de la app puede ver con un solo clic qué categorías (como Peluches o Accesorios) sí están dejando dinero limpio y cuáles no.

En conclusión, el ETL fue como el filtro que limpió el agua sucia. Ahora que el dataset maestro `df_df` está impecable, bien relacionado y exportado a CSV, estamos listos para hacer el Análisis Exploratorio (EDA) en un notebook aparte, con la total seguridad de que los resultados son 100% reales y confiables.

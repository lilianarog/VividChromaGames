```python
ventas_municipio = df_df.groupby('municipio').agg(
    Total_Unidades=('cantidad', 'sum'),
    Total_Ingresos=('Ingreso_Total', 'sum'),
    Utilidad_Total=('Utilidad', 'sum')
).reset_index().sort_values(by='Total_Unidades', ascending=False)

mun_mas = ventas_municipio.iloc[0]
mun_menos = ventas_municipio.iloc[-1]

print("🏆 MUNICIPIO CON MÁS COMPRAS:", mun_mas['municipio'], f"({mun_mas['Total_Unidades']} unidades)")
print("📉 MUNICIPIO CON MENOS COMPRAS:", mun_menos['municipio'], f"({mun_menos['Total_Unidades']} unidades)")
print("\n--- RESUMEN COMPLETO DE MUNICIPIOS ---")
print(ventas_municipio)

# Gráfico de Barras de Municipios
plt.figure(figsize=(9, 4))
sns.barplot(data=ventas_municipio, x='municipio', y='Total_Unidades', palette='magma')
plt.title('Unidades Vendidas por Municipio de Puebla', fontsize=12, fontweight='bold')
plt.xlabel('Municipio')
plt.ylabel('Cantidad de Unidades')
plt.tight_layout()
plt.show()
```

```python
top_productos = df_df.groupby('nombre').agg(
    Unidades_Vendidas=('cantidad', 'sum'),
    Utilidad=('Utilidad', 'sum')
).reset_index().sort_values(by='Unidades_Vendidas', ascending=False)

print("--- LO QUE MÁS SE VENDE (PRODUCTOS TOP) ---")
print(top_productos)

# Gráfico de Productos
plt.figure(figsize=(9, 4))
sns.barplot(data=top_productos, x='Unidades_Vendidas', y='nombre', palette='viridis')
plt.title('Productos Más Vendidos', fontsize=12, fontweight='bold')
plt.xlabel('Unidades Vendidas')
plt.ylabel('Producto')
plt.tight_layout()
plt.show()
```
